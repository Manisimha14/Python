```python
import numpy as np
```

Creating Array from List


```python
# 1d array
one_d=np.array([1,2,3])
two_d=np.array([[1,2,3],[4,5,6]])
print(one_d)
print(two_d)

```

Difference Between List and Numpy Array

```python
listmultiplication=[1,2,3]
nparraymultiplication=np.array([1,2,3])
print(listmultiplication*2)
print(nparraymultiplication*2)
# this is broadcasting nothing but expanding the scalar and performing the calculations
```

```python
print(np.arange(10))
# this will gives from 0 to n-1
# and the parameters are start,stop and step
```

Creating Array from Scratch

```python
zeros=np.zeros([3,4])
# this will generate the zeros 2d array of 3 rows and 4 columns
print(zeros)
ones=np.ones([3,4])
print(ones)
full=np.full([3,4],69)
# this is used to generate the matrix of specified value
print(full)
random=np.random.random((3,4))
print(random)
sequence=np.arange(0,10,3)
print(sequence)
```

### Vector Matrix and Tensor

```python
vector=np.array([1,2,3])
matrix=np.array([[1,2,3],[4,5,6]])
tensor=np.array([[[1,2,3],[4,5,6]]])
# multi dimensional array is known as tensors
print(vector)
print(matrix)
print(tensor)
```

### Array Properties

```python
array=np.array([[1,2,3],[4,5,6]])
print("shape",array.shape)
# gives rows and columns
print("dimension",array.ndim)
print("size",array.size)
print("dtype",array.dtype)
```

### Array Reshaping


```python
array1= np.arange(12)
reshaped=array1.reshape(3,4)
print("\n reshaped array",reshaped)
flatten=reshaped.flatten()
print("\n flatten array",flatten)
ravel=reshaped.ravel()
print("\n ravel array",ravel)
# ravel returns the orginal instead of copy
transposed=reshaped.T
print("\n Transposed array",transposed)
```
