```python
data=[100,200,300]
series=pd.Series(data,index=["a","b","c"])
print(series)
# for accesing the data using the label
print(series.loc["a"])
print(series.iloc[1])
```

```python
# filtering the data
data2=[100,200,300,400,500]
series2=pd.Series(data2,index=["a","b","c","d","e"])
series2.loc["f"]=600

print(series2[series2>200])

```

```python
calories={
    "Day1":100,
    "Day2":200,
    "Day3":300,
    "Day4":400,
}
series=pd.Series(calories)
series.loc["Day5"]=500
print(series[series>200])
```

```python
data={
    "Name":["Mani","Gani","Devi"],
    "Age":[10,20,30]
}
df=pd.DataFrame(data,index=["Person1","Person2","Person3"])
# adding a new column
df["Gender"]=["M","M","F"]
# adding a new row
new_row=pd.DataFrame([{"Name":"Narayana","Age":40,"Gender":"M"}],index=["Person4"])
new_df=pd.concat([df,new_row])
print(new_df)


print(df.iloc[0])
```

```python
df=pd.read_csv("data.csv",index_col="Name")
print(df)
```

```python
# Selection
# selection by column
# df["Name"]
# and this is a series
df["Height"]
# df[["Name","Height"]]
# Selection By Rows
df.loc["Pikachu",["Height","Weight"]]
#for only particular columns
# We can also use slicing method
df.iloc[0:11,[2,3]]
```

```python
# Keeping the rows which match the given condition
tall_pokemon=df[df["Height"]>3]
print(type(tall_pokemon))
legendary=df[df["Legendary"]==1]
print(legendary)
```

```python
# AGGREGATION
print(df.mean(numeric_only=True))
print(df.sum(numeric_only=True))
print(df["Height"].mean())
```

```python
group=df.groupby("Type1")
group["Height"].mean()
```

```python
# Data Cleaning

df=df.dropna(subset=["Type2"])
df=df.fillna({"Type1":"None"})
df.drop_duplicates()
print(df)
```
