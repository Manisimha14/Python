# 📊 Data Visualization Summary

---

## 📈 Histogram

* **Type:** Univariate Analysis
* **Purpose:** Displays the distribution of a continuous variable.
* **Example:** Students' marks distribution.

---

## 📊 Bar Graph

* **Type:** Univariate / Bivariate Analysis
* **Purpose:** Compares values across different categories.
* **Example:** Branch vs. Average Package.

---

## 🥧 Pie Chart

* **Type:** Univariate Analysis
* **Purpose:** Represents the percentage contribution of each category.
* **Example:** Branch-wise student distribution.

---

## 📦 Box Plot

* **Type:** Univariate Analysis
* **Purpose:** Identifies outliers and summarizes data distribution.
* **Shows:**

  * Minimum
  * Q1 (25th Percentile)
  * Median (50th Percentile)
  * Q3 (75th Percentile)
  * Maximum
  * Interquartile Range (IQR)
  * Potential Outliers

---

## ✨ Scatter Plot

* **Type:** Bivariate Analysis
* **Purpose:** Examines the relationship between two continuous variables.
* **Example:** CGPA vs. Average Package.

---

> **Quick Tip:**
> **Histogram → Distribution**
> **Bar Graph → Comparison**
> **Pie Chart → Proportion**
> **Box Plot → Outliers & Spread**
> **Scatter Plot → Relationship**



# 📊 Selection of Visualization Techniques

| Plot Type      | Analysis Type          | Data Type   | Best Used For                                   |
| :------------- | :--------------------- | :---------- | :---------------------------------------------- |
| 📈 Histogram   | Univariate             | Continuous  | Distribution of numerical data                  |
| 📊 Bar Graph   | Univariate / Bivariate | Categorical | Comparing categories                            |
| 🥧 Pie Chart   | Univariate             | Categorical | Showing proportions or percentages              |
| 📦 Box Plot    | Univariate             | Continuous  | Detecting outliers and summarizing distribution |
| ✨ Scatter Plot | Bivariate              | Continuous  | Finding relationships between two variables     |

---

# 📚 Data Classification

| Analysis Type  |    Categorical Data    |    Continuous Data    |
| :------------- | :--------------------: | :-------------------: |
| **Univariate** | ✅ Bar Graph, Pie Chart | ✅ Histogram, Box Plot |
| **Bivariate**  |   ✅ Grouped Bar Graph  |     ✅ Scatter Plot    |

---

### 🔑 Quick Rule

| Data Type       | Univariate           | Bivariate         |
| :-------------- | :------------------- | :---------------- |
| **Categorical** | Bar Graph, Pie Chart | Grouped Bar Graph |
| **Continuous**  | Histogram, Box Plot  | Scatter Plot      |

# 📦 Box Plot (Box-and-Whisker Plot)

## 📌 Overview

* **Data Type:** Continuous
* **Analysis Type:** Univariate
* **Purpose:** Summarizes the distribution of data and helps identify outliers.

---

## 📚 Components

| Component                         | Description                                           |
| :-------------------------------- | :---------------------------------------------------- |
| **Minimum**                       | Smallest non-outlier value                            |
| **Q1 (25th Percentile)**          | 25% of observations lie below this value              |
| **Median (Q2 - 50th Percentile)** | Middle value of the dataset                           |
| **Q3 (75th Percentile)**          | 75% of observations lie below this value              |
| **IQR (Interquartile Range)**     | `IQR = Q3 - Q1`                                       |
| **Whiskers**                      | Extend to the smallest and largest non-outlier values |
| **Outliers**                      | Data points lying beyond `1.5 × IQR`                  |

---

## 📏 Whisker Rule

```text
Lower Limit = Q1 - 1.5 × IQR
Upper Limit = Q3 + 1.5 × IQR
```

* Values within these limits are included in the whiskers.
* Values outside these limits are plotted as **outliers**.

---

## 🔍 Interpretation

✅ **Small Box** → Low variability

✅ **Large Box** → High variability

✅ **Median at Center** → Approximately symmetric data

✅ **Long Upper Whisker** → Right-skewed distribution

✅ **Long Lower Whisker** → Left-skewed distribution

✅ **Points Outside Whiskers** → Potential outliers

---

## 📝 Key Takeaways

* The **box contains the middle 50% of the data**.
* The **line inside the box represents the median**.
* The **whiskers show the normal spread of the data**.
* **Outliers are displayed as individual points**.

> **Remember:**
> **Box → Middle 50% (IQR)**
> **Line → Median**
> **Whiskers → Normal Range**
> **Dots → Outliers**


````python
sns.boxplot(data=data,x="Global_Sales")
````

## 📦 Structure of a Box Plot

```text
Lower      Q1      Median      Q3       Upper
Whisker                             Whisker
    |---------|=======|=======|------------|
                                        ● ● ● ●
                                        Outliers
```

## 📖 Legend

| Symbol            | Meaning                           |
| :---------------- | :-------------------------------- |
| **Lower Whisker** | Smallest non-outlier value        |
| **Q1**            | First Quartile (25th Percentile)  |
| **Median**        | Second Quartile (50th Percentile) |
| **Q3**            | Third Quartile (75th Percentile)  |
| **Upper Whisker** | Largest non-outlier value         |
| **●**             | Outliers                          |

# 🎨 Seaborn Count Plot — Visual Intuition Guide

> A Count Plot answers one simple question:
>
> **"How many times does each category appear?"**

---

# 🌟 Think of Count Plot as a Frequency Counter

Suppose we have:

```python
Gender = [
    "Male",
    "Female",
    "Male",
    "Male",
    "Female"
]
```

Count Plot internally performs:

```python
Male   → 3
Female → 2
```

And visualizes it as:

```text
Male      ███

Female    ██
```

---

# 📦 data

## What does it do?

Tells Seaborn **where the data lives**.

```python
sns.countplot(
    data=df,
    x="Gender"
)
```

Without `data`, Seaborn doesn't know where to find the column.

### Mental Model

```text
data = The Library 📚
x    = Which Book to Read 📖
```

---

# 📍 x

## What does it do?

Places categories on the horizontal axis.

```python
sns.countplot(
    data=df,
    x="Gender"
)
```

### Visualization

```text
Count
 ↑

3 |       █
2 |       █       █
1 |       █       █
0 +----------------------→

       Male   Female
```

### Use When

You want normal vertical bars.

---

# 📍 y

## What does it do?

Places categories on the vertical axis.

```python
sns.countplot(
    data=df,
    y="Gender"
)
```

### Visualization

```text
Male      ███

Female    ██
```

### Use When

Category names are long.

Example:

```text
Computer Science Engineering
Electronics and Communication
Mechanical Engineering
```

Horizontal bars become easier to read.

---

# 🎭 hue

## What does it do?

Splits each category into smaller colored groups.

```python
sns.countplot(
    data=df,
    x="Gender",
    hue="Survived"
)
```

### Without hue

```text
Male      ████

Female    ██
```

Only total counts.

---

### With hue

```text
Male

█ █

Female

█ █
```

Now we see:

```text
Male Survived
Male Not Survived

Female Survived
Female Not Survived
```

### Superpower

Instead of answering:

> How many males?

It answers:

> How many surviving males?
>
> How many non-surviving males?

---

# 📋 order

## What does it do?

Controls category order.

### Default

```text

```

---

```python
sns.countplot(
    data=df,
    x="Grade",
    order=["A","B","C","D"]
)
```

Produces:

```text
A B C D
```

### Use When

Natural ordering exists.

Examples:

```text
Poor
Average
Good
Excellent
```

or

```text
Freshman
Sophomore
Junior
Senior
```

# 🎯 When Should You Use Count Plot?

Use Count Plot when:

```text
Gender
Department
Grade
Country
Passenger Class
Survival Status
Product Type
Source
Category
```

In short:

> Whenever the data is CATEGORICAL.

---

# ❌ When NOT to Use Count Plot

Don't use Count Plot for:

```text
Age
Height
Weight
Salary
Fare
Temperature
Marks
```

These are continuous variables.

Use:

```python
histplot()
boxplot()
violinplot()
kdeplot()
```

instead.
