```python
# factorial
n=int(input())
ans=n
if(n==0):
    print(1)
while(n!=1):
    ans=ans*(n-1)
    n=n-1
print(ans)    
```

```python
# reverse the digits
n=int(input())
copy=n
absn=abs(n)
copyneg=absn
if(n>=0):
    reverse=n%10
    n=n//10
    while(n!=0):
     r=n%10
     n=n//10
     reverse=reverse*10+r
    if(copy==reverse):
      print(True)   
    else:
      print(False)  

else:
   reverse=absn%10
   absn=absn//10
   while absn!=0:
    r=absn%10
    absn=absn//10
    reverse=reverse*10+r
   if(copyneg==reverse):
      print(True)
   else:
      print(False) 
```
