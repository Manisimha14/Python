````python
import pandas as pd
def select_rows(df, r1, r2):
    '''
    input:
    df -> the dataframe provided to the function
    r1 -> the starting index of the rows to be selected
    r2 -> the ending index of the rows to be selected
    
    output:
    df_new -> the selected rows of the dataframe
    '''
    
    # Select the rows r1 to r2
    df_selected = df.iloc[r1:r2,:]
    
    # Select the columns "name" and "amt"
    df_new = df_selected.loc[:,["name","amt"]]
    
    return df_new
````

# 🔍 Pandas `unique()` vs `nunique()`

Understanding the difference between **`unique()`** and **`nunique()`** is essential for effective data exploration and cleaning.

---

## 🎯 1. `unique()`

Returns the **distinct values** present in a Series.

### Example

```python
import pandas as pd

s = pd.Series([1, 1, 2, 3, 3])

s.unique()
```

### Output

```python
array([1, 2, 3])
```

### Key Point

✅ Returns the **actual unique values**

❌ Not available directly on a DataFrame

---

## 🎯 2. `nunique()`

Returns the **count of unique values**.

### Example

```python
s.nunique()
```

### Output

```python
3
```

### Key Point

✅ Returns the **number of distinct values**

---

# 📊 DataFrame Behavior

## ❌ `df.unique()`

```python
df.unique()
```

### Result

```text
AttributeError:
'DataFrame' object has no attribute 'unique'
```

`unique()` works only on a **Series**.

---

## ✅ `df.nunique()`

```python
df.nunique()
```

### Output

```python
A    3
B    5
C    2
dtype: int64
```

Each value represents the number of unique values in that column.

---

# 📝 Statement Evaluation

| Statement | Verdict |
|------------|---------|
| `.unique()` method returns unique values in a Series | ✅ True |
| `.unique()` on a DataFrame returns unique values for each column | ❌ False |
| `.nunique()` method returns the count of unique elements in a Series | ✅ True |
| `.nunique()` on a DataFrame returns the count of unique elements for each column | ✅ True |

---

```text
    └── Returns COUNT

    └── Returns COUNT per Column
```

# 📍 Pandas `loc` vs `iloc` — Complete Guide

Understanding the difference between **`loc`** and **`iloc`** is one of the most important concepts in Pandas.

---

## 🗂️ Sample DataFrame

| Index | Name   | Gender | Profession |
|---------|--------|---------|------------|
| 0 | Jim    | M | Athlete |
| 1 | Carrie | F | Tech |
| 2 | Chris  | M | Cricketer |
| 3 | Morris | M | Actor |

---

# 🔹 Line A

```python
df.iloc[:2, :2]
```

### How `iloc` Works

`iloc` uses **integer positions**.

```text
Rows    → 0 to 1
Columns → 0 to 1
```

### Selected Rows

```text
0, 1
```

### Selected Columns

```text
Name, Gender
```

### Output

| Name | Gender |
|--------|--------|
| Jim | M |
| Carrie | F |

---

## 🎯 Conclusion for Line A

❌ Statement 1 is False

> It does NOT include the `Profession` column.

✅ Statement 2 is True

> It returns the first two rows and the first two columns (`Name`, `Gender`).

---

# 🔹 Line B

```python
df.loc[:2, "Name":"Profession"]
```

### How `loc` Works

`loc` uses **labels**.

Unlike `iloc`, **the ending label is included**.

---

### Selected Rows

```python
:2
```

returns:

```text
0, 1, 2
```

---

### Selected Columns

```python
"Name":"Profession"
```

returns:

```text
Name
Gender
Profession
```

because `loc` includes the ending column label.

---

### Output

| Name | Gender | Profession |
|--------|--------|------------|
| Jim | M | Athlete |
| Carrie | F | Tech |
| Chris | M | Cricketer |

---

## 🎯 Conclusion for Line B

✅ Statement 3 is True

> Rows 0, 1, and 2 with columns `Name`, `Gender`, and `Profession` are returned.

❌ Statement 4 is False

> No error is generated.

---

# ⚡ loc vs iloc

| Feature | `loc` | `iloc` |
|----------|---------|---------|
| Selection Type | Labels | Integer Positions |
| End Index Included | ✅ Yes | ❌ No |
| End Column Included | ✅ Yes | ❌ No |
| Example | `df.loc[:2]` → 0,1,2 | `df.iloc[:2]` → 0,1 |

---

```text
loc
```

# 🔍 Understanding `apply()` with `axis=1` and Missing Values

Consider the function:

```python
def is_null(x):
    return sum(x.isnull())
```

---

## 🧠 What Does This Function Do?

For any Series `x`:

```python
x.isnull()
```

returns a Boolean mask:

```python
[False, True, False, True]
```

where:

- `True` → Missing Value (`NaN`)
- `False` → Non-Missing Value

Then:

```python
sum(x.isnull())
```

counts the number of `True` values.

### Example

```python
x = pd.Series([10, None, 20, None])

x.isnull()
```

Output:

```python
[False, True, False, True]
```

```python
sum(x.isnull())
```

Output:

```python
2
```

✅ There are **2 missing values**.

---

# 📍 Understanding `apply()`

Given:

```python
df.apply(is_null, axis=1)
```

The `apply()` function executes `is_null()` on each row because:

```python
axis = 1
```

means:

```text
Apply Row-wise
```

---

## 🎯 Visual Representation

```text
DataFrame
│
├── Row 0 → is_null()
├── Row 1 → is_null()
├── Row 2 → is_null()
└── Row n → is_null()
```

Each row is passed as a Series to the function.

---

### Input DataFrame

|---|---|---|
| 1 | NaN | 3 |
| NaN | 5 | NaN |
| 7 | 8 | 9 |

```python
df.apply(is_null, axis=1)
```

# 🎨 What is Happening Internally?

```python

result = []

for each_row in df:
    result.append(sum(each_row.isnull()))
```

Result:


# ⚡ axis = 0 vs axis = 1

| Axis | Operation | Direction |
|--------|------------|------------|
| `axis=0` | Column-wise | ↓ Vertical |
| `axis=1` | Row-wise | → Horizontal |

---

### Column-wise

```python

```

Output:

```python
A    1
B    1
C    1
```

Counts missing values **per column**.

---

### Row-wise

```python
df.apply(is_null, axis=1)
```

Output:

```python
0    1
1    2
2    0
```

Counts missing values **per row**.

---

# 🚀 Memory Trick

```text
axis = 0
   ↓
Columns


axis = 1
   →
Rows
```

# 🔍 Understanding `isin()` in Pandas

The `isin()` method is used to check whether values exist in a given collection.

---

## 📌 Sample DataFrame

```python
import pandas as pd

data = {
    'cust_id': [101, 102, 103, 104],
    'name': ['rick', 'morty', 'pickle', 'jerry']
}

df = pd.DataFrame(data)
```

### DataFrame

| Index | cust_id | name |
|--------|---------|--------|
| 0 | 101 | rick |
| 1 | 102 | morty |
| 2 | 103 | pickle |
| 3 | 104 | jerry |

---

# 🎯 Option 1

```python
df.isin([101, 102, 120])
```

### What Happens?

Checks every cell in the DataFrame against:

```python
[101, 102, 120]
```

### Output

| cust_id | name |
|----------|----------|
| True | False |
| True | False |
| False | False |
| False | False |

✅ **Valid Syntax**

---

# 🎯 Option 2

```python
df.isin({'cust_id': [101, 102]})
```

### What Happens?

Checks only the specified column.

### Output

| cust_id | name |
|----------|----------|
| True | False |
| True | False |
| False | False |
| False | False |

✅ **Valid Syntax**

---

# 🎯 Option 3

```python
df.isin(df.cust_id)
```

### What Happens?

`df.cust_id` is a Series:

```python
0    101
1    102
2    103
3    104
Name: cust_id
```

Pandas can use a Series as input to `isin()`.

### Output

| cust_id | name |
|----------|----------|
| True | False |
| True | False |
| True | False |
| True | False |

✅ **Valid Syntax**

---

# ❌ Option 4

```python
df.name.isin('rick')
```

### Error

```python
TypeError:
only list-like objects are allowed to be passed to isin(),
you passed a 'str'
```

### Why?

`isin()` expects a **collection** of values.

Valid inputs:

```python
['rick']
('rick',)
{'rick'}
pd.Series(['rick'])
np.array(['rick'])
```

But:

```python
'rick'
```

is just a single string, not a collection.

Therefore:

```python
df.name.isin('rick')
```

❌ Raises an Error

---

# 🚀 Correct Way

### Single Value

```python
df.name.isin(['rick'])
```

Output:

```python
0     True
1    False
2    False
3    False
Name: name, dtype: bool
```

---

### Multiple Values

```python
df.name.isin(['rick', 'jerry'])
```

Output:

```python
0     True
1    False
2    False
3     True
Name: name, dtype: bool
```

---

# 📊 Visual Summary

```text
isin()
│
├── List      ✅
├── Tuple     ✅
├── Set       ✅
├── Series    ✅
├── Array     ✅
│
└── String    ❌
```

---

# 🧠 Memory Trick

Think of `isin()` as asking:

> "Is this value contained inside THIS COLLECTION?"

So always provide a collection:

### ❌ Wrong

```python
df.col.isin('A')
```

### ✅ Correct

```python
df.col.isin(['A'])
```

Even if you're checking a single value, wrap it in a list.

---


because `isin()` requires a **list-like object**, not a string.
