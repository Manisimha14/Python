### Defective Product Given Machine Source
### A factory has two machines:
### Machine A produces 60% of items and has defect rate 2%
### Machine B produces 40% of items and has defect rate 5%

### Simulate 300,000 produced items using NumPy.

### Let:
### D = item is defective
### B = item came from Machine B

### Estimate the conditional probability:

### P(D | B)

```python
machines=np.array(["A","A","A","A","A","A","B","B","B","B"])
n=100000
random=np.random.choice(machines,n)
defective=np.zeros(n,dtype=bool)
defective[random=="A"]=np.random.rand(np.sum(random=="A")) <0.02
defective[random=="B"]=np.random.rand(np.sum(random=="B")) <0.05
B=random=="B"
print(np.sum(defective & B)/np.sum(B))
```

 A bank has implemented a fraud detection system.

- 2% of all transactions are fraudulent.
- The system correctly flags 92% of fraudulent transactions.
- The system incorrectly flags 4% of legitimate transactions as fraud.

A transaction is flagged as fraudulent.

Questions:

1. Compute theoretically:
   P(Fraud | Flagged)


```python
fradulent=np.random.rand(1000) <0.02
flags=np.zeros(1000 ,dtype=bool)
flags[fradulent]=np.random.rand(np.sum(fradulent)) <0.92
flags[~fradulent]=np.random.rand(np.sum(~fradulent)) <0.04
print(np.sum((fradulent) & (flags))/np.sum(flags))
```

An email system classifies emails as spam.

- 35% of emails are spam.
- 85% of spam emails contain the word "DISCOUNT".
- 6% of non-spam emails contain the word "DISCOUNT".

You receive an email that contains the word "DISCOUNT".
 Compute theoretically:
   P(Spam | Contains DISCOUNT)


```python
n=10000
spam=np.random.rand(n) <0.35
discountcontain=np.zeros(n,dtype=bool)
discountcontain[spam]=np.random.rand(np.sum(spam)) <0.85
discountcontain[~spam]=np.random.rand(np.sum(~spam)) <0.06
print(np.sum(spam & discountcontain)/np.sum(discountcontain))
```

Three machines produce light bulbs:

- Machine A produces 50% of total bulbs and has a 2% defect rate.
- Machine B produces 30% of total bulbs and has a 4% defect rate.
- Machine C produces 20% of total bulbs and has a 9% defect rate.

A randomly selected bulb is defective.
 Compute theoretically:
   P(Bulb came from Machine C | Defective)


```python
machines=np.random.choice(["A","B","C"],n,p=[0.5,0.3,0.2])
defective=np.zeros(n,dtype=bool)
defective[machines=="A"]=np.random.rand(np.sum(machines=="A")) <0.02
defective[machines=="B"]=np.random.rand(np.sum(machines=="B")) <0.04
defective[machines=="C"]=np.random.rand(np.sum(machines=="C")) <0.09
machinec=machines=="C"
print(np.sum(defective & machinec)/np.sum(defective))
```
