### Basic Concepts in Probability
### 🌟 <font color="skyblue">1. Experiment, Outcomes, and Sample Space</font>

### **Experiment**  
An experiment is any action or process that generates outcomes.  
💡 *Example*: Observing Sachin’s performance in an ODI match is an experiment.

### **Outcome**  
An outcome is a single result of an experiment.  
💡 *Example*: Sachin scoring 100 runs in a specific match is one possible outcome.

### **Sample Space (S)**  
The sample space is the set of all possible outcomes of an experiment.  
💡 *Example*: For Sachin’s ODI career, the sample space might include all possible scores he could achieve in a match (0, 1, 2, ..., 200+).

```python
sample_space=data["runs"].unique()
display(sample_space)
```

### 🌟 <font color="green">2. Events</font>

An event is a subset of the sample space. It can contain one or more outcomes.  
💡 *Example*:  
- **Event A**: Sachin scores more than 100 runs in a match.  
- **Event B**: Sachin scores exactly 0 runs (a duck).

### **Types of Events**

1. **Mutually Exclusive Events**  
   Events that cannot happen at the same time.  
   💡 *Example*: Sachin scoring a century and scoring a duck in the same match.

2. **Joint Events**  
   Events that can happen simultaneously.  
   💡 *Example*: Sachin scores more than 50 runs *and* the match is won by India.

3. **Independent Events**  
   Events where the occurrence of one does not affect the other.  
   💡 *Example*: The outcome of Sachin’s performance and the weather conditions (assuming no match cancellations).

4. **Exhaustive Events**  
   A set of events that covers all possible outcomes in the sample space.  
   💡 *Example*:  
   - Event A: Sachin scores less than 50 runs.  
   - Event B: Sachin scores 50 or more runs.  
   Together, these two events are exhaustive.

```python
## Which ODI matches did Sachin Tendulkar score more than 50 runs and India won, and what were the details of those matches?
fiftyruns=data[data["runs"]>50]
indiawon=data[data["Won"]=="True"]
intersection=pd.merge(fiftyruns,indiawon,how="inner")
display(intersection)
```

### For Z_Score Problems

Suppose your city has a large population, where the average height is 65 inches with a standard deviation of 4 inches, and your height is 60 inches.

What is the probability that a randomly selected person in your city is shorter than you?
(Round off the answer to 4 decimal points)

```python
from scipy.stats import norm
meanheight=65
sd=4
myheight=60
zscore=(myheight-meanheight)/sd
prob=norm.cdf(zscore)
display(np.round(prob,4))
```

### For Exponential Distribution Problems

```python
from scipy.stats import expon
mean=2
atmost6=expon.cdf(6,scale=mean)
lesstahn6=1-expon.cdf(6,scale=mean)
display(atmost6,lesstahn6)
```

The amount of money customers spend in one trip to the supermarket follows an exponential distribution with a mean amount of Rs 1000.

If a customer has Rs 600 in the wallet, what is the probability that the customer will need more money?

Note: Round the output up to 2 decimal places.

```python
from scipy.stats import expon
mean=1000
prob=1-expon.cdf(600,scale=mean)
display(np.round(prob,2))
```
