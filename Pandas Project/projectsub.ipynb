### Importing The Data set

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
url = "https://d2beiqkhq929f0.cloudfront.net/public_assets/assets/000/001/125/original/aerofit_treadmill.csv"

df = pd.read_csv(url)

df.head()
```

### Information about the data set
Product info about the features of the Dataset:

---

##  Features Description

| Feature | Description |
|--------|------------|
| **Product Purchased** | Type of treadmill purchased: KP281, KP481, or KP781 |
| **Age** | Age of the customer (in years) |
| **Gender** | Gender of the customer (Male/Female) |
| **Education** | Years of education completed |
| **Marital Status** | Customer's marital status (Single/Partnered) |
| **Usage** | Average number of times the customer plans to use the treadmill per week |
| **Income** | Annual income of the customer (in USD) |
| **Fitness** | Self-rated fitness level (1 = Poor, 5 = Excellent) |
| **Miles** | Average number of miles the customer expects to walk/run per week |

---

---

##  Product Portfolio

| Product | Category | Price | Description |
|--------|----------|--------|-------------|
| **KP281** | Entry-Level | $1,500 | Suitable for beginners and budget-conscious users |
| **KP481** | Mid-Level | $1,750 | Designed for regular fitness users and intermediate runners |
| **KP781** | Premium | $2,500 | Advanced treadmill with high-end features for serious fitness enthusiasts |

### Outlier Detection in Univariate Analysis

```python
numericalcolumns = ['Age', 'Education', 'Usage', 'Income', 'Fitness', 'Miles']
categoricalcoulmns = ['Product', 'Gender', 'MaritalStatus']


for col in numericalcolumns:
    plt.figure(figsize=(6,4))
    sns.boxplot(df[col])
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
numericalcolumns = ['Age', 'Education', 'Usage', 'Income', 'Fitness', 'Miles']
for col in numericalcolumns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

### Outlier Detection in Bivariate Analysis

Outliers in the case of product and income

```python
for product in df['Product'].unique():
    subset = df[df['Product'] == product]

    Q1 = subset['Income'].quantile(0.25)
    Q3 = subset['Income'].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = subset[(subset['Income'] < lower) | (subset['Income'] > upper)]

    print(f"{product}: {len(outliers)} outliers")
```

Outlier in the case of product and fitness

```python
for product in df["Product"].unique():
    subset=df[df["Product"]==product]
    q1=subset["Fitness"].quantile(0.25)
    q3=subset["Fitness"].quantile(0.75)
    iqr=q3-q1
    lower = Q1 - 1.5 * iqr
    upper = Q3 + 1.5 * iqr
    outliers=subset[(subset["Fitness"]<lower) |(subset["Fitness"]>upper) ]
    print(f"{product}:{len(outliers)} outliers")
```

```python
for product in df["Product"].unique():
    subset=df[df["Product"]==product]
    q1=subset["Usage"].quantile(0.25)
    q3=subset["Usage"].quantile(0.75)
    iqr=q3-q1
    lower = Q1 - 1.5 * iqr
    upper = Q3 + 1.5 * iqr
    outliers=subset[(subset["Usage"]<lower) |(subset["Usage"]>upper) ]
    print(f"{product}:{len(outliers)} outliers")
```
